```python

import re
import json
import pandas as pd
from pathlib import Path

# ==========================================
# 1. LOAD ORIGINAL PILOT OUTPUT
# ==========================================

BASE_DIR = Path.cwd()

RAW_OUTPUT_FILE = (
    BASE_DIR / "results" / "qwen" / "pilot_output_raw.txt"
)

raw_text = RAW_OUTPUT_FILE.read_text(encoding="utf-8")

# ==========================================
# 2. REBUILD PILOT SAMPLE ORDER
# ==========================================

if "pilot_samples" not in globals():

    DATASET_FILE = (
        BASE_DIR / "malay_cleaned_dataset"
        / "malay_evaluation_dataset.jsonl"
    )

    records = []

    with open(DATASET_FILE, "r", encoding="utf-8") as f:
        for line in f:
            records.append(json.loads(line))

    df = pd.DataFrame(records)

    pilot_samples = (
        df.groupby("domain", group_keys=False)
          .sample(n=5, random_state=42)
          .reset_index(drop=True)
    )

# ==========================================
# 3. PARSE RECOVERED OUTPUT
# ==========================================

pattern = re.compile(
    r"\[(\d+)/25\]\s*([^\r\n]+)\s*"
    r"Status:\s*(success|failed)\s*"
    r"Time:\s*([\d.]+)\s*seconds\s*"
    r"(?:Prediction:\s*(.*?))?"
    r"(?=\n\s*\[\d+/25\]|\Z)",
    re.IGNORECASE | re.DOTALL
)

matches = list(pattern.finditer(raw_text))

recovered_results = []

for match in matches:

    pilot_id = int(match.group(1))
    domain = match.group(2).strip()
    status = match.group(3).lower()
    duration = float(match.group(4))

    prediction = match.group(5)

    if prediction:
        prediction = prediction.strip()
    else:
        prediction = None

    if pilot_id < 1 or pilot_id > len(pilot_samples):
        continue

    record = pilot_samples.iloc[pilot_id - 1].to_dict()

    # Verify domain alignment
    if record["domain"] != domain:
        print(f"WARNING: Domain mismatch at pilot {pilot_id}")
        continue

    recovered_results.append({
        "pilot_id": pilot_id,
        "domain": domain,
        "input": record["input"],
        "reference": record["reference"],
        "prediction": prediction,
        "inference_time": duration,
        "status": status,
        "error": None
    })

# ==========================================
# 4. SAVE RECOVERED RESULTS
# ==========================================

RESULT_DIR = BASE_DIR / "results" / "qwen"
RESULT_DIR.mkdir(parents=True, exist_ok=True)

PILOT_FILE = RESULT_DIR / "pilot_results.jsonl"

with open(PILOT_FILE, "w", encoding="utf-8") as f:
    for result in recovered_results:
        f.write(json.dumps(result, ensure_ascii=False) + "\n")

# ==========================================
# 5. VERIFY RECOVERY WITHOUT SHOWING DATA
# ==========================================

pilot_df = pd.DataFrame(recovered_results)

print("=" * 55)
print("PILOT RECOVERY AUDIT")
print("=" * 55)

print("Recovered records:", len(pilot_df))
print("Unique pilot IDs:", pilot_df["pilot_id"].nunique())
print("Successful:", (pilot_df["status"] == "success").sum())
print("Failed:", (pilot_df["status"] == "failed").sum())
print("Missing predictions:", pilot_df["prediction"].isna().sum())
print("Missing references:", pilot_df["reference"].isna().sum())

print("\nDomain distribution:")
print(pilot_df["domain"].value_counts())

print("\nAverage inference time:",
      round(pilot_df["inference_time"].mean(), 2), "seconds")

print("\nSaved file:", PILOT_FILE)
print("File exists:", PILOT_FILE.exists())
```
